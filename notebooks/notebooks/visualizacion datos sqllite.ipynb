```python
import sqlite3
import pandas as pd
from pathlib import Path

# Detectar la raíz del proyecto para encontrar el 'mlflow.db' real
current_dir = Path(".").resolve()
PROJECT_ROOT = current_dir

for parent in [current_dir] + list(current_dir.parents):
    if (parent / "src").exists():
        PROJECT_ROOT = parent
        break

db_path = PROJECT_ROOT / "mlflow.db"
print(f"Conectando a la base de datos real en: {db_path}")

# Conectar al archivo SQLite del proyecto
conn = sqlite3.connect(db_path)

# 1. Listar todas las tablas generadas por MLflow
tables = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table';", conn)
print("\n--- Tablas encontradas en MLflow ---")
print(tables['name'].tolist())

# 2. Consultar los experimentos registrados
df_experiments = pd.read_sql_query("SELECT * FROM experiments;", conn)
print("\n--- Experimentos ---")
print(df_experiments[['experiment_id', 'name', 'artifact_location']])

# 3. Consultar las corridas (runs)
df_runs = pd.read_sql_query("SELECT run_uuid, experiment_id, status, start_time FROM runs;", conn)
print("\n--- Ultimas corridas (Runs) ---")
print(df_runs.head())

# 4. Consultar las métricas registradas en la base de datos
df_metrics = pd.read_sql_query("SELECT * FROM metrics LIMIT 10;", conn)
print("\n--- Muestra de la tabla metrics ---")
print(df_metrics)

conn.close()
```
